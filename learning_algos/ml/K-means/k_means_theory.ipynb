## Drawback of K-means

1. Need to specify K
2. Sensitive to initialization
3. Assumes roughly spherical clusters 
3. sensitive to outliers 
5. feature with larse numrical ranges dominates 



## How to solve these limitations 

| Problem | Possible solution |
| :--- | :--- |
| Unknown K | Elbow, Silhouette, Davies-Bouldin, CHI |
| Initialization sensitivity | K-Means++ and multiple initializations (n_init) |
| Non-spherical clusters | DBSCAN, Spectral Clustering |
| Outliers | Remove/treat outliers or use K-Medoids |
| Varying cluster densities | HDBSCAN or DBSCAN |
| Feature scaling | StandardScaler or RobustScaler |
| Categorical data | K-Modes or K-Prototypes |
| Need probability assignments | Gaussian Mixture Models (GMM) |

## Time Complexity: DBSCAN vs K-Means

| Algorithm | Average / Typical | Worst Case |
| :--- | :--- | :--- |
| K-Means | $O(N \cdot K \cdot D \cdot I)$ | No simple polynomial bound on iterations in general |
| DBSCAN | $O(N\log N)$ with efficient spatial indexing in suitable low-dimensional data | $O(N^2)$ |

Where:

- $N$ = Number of observations
- $K$ = Number of clusters
- $D$ = Number of features/dimensions
- $I$ = Number of K-Means iterations
