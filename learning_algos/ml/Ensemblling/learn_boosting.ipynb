# Boosting Algos

Boosting algorithms are designed to have low bias and low variance by combining many simple, high-bias models (weak learners) sequentially while using mathematical and structural safeguards to control complexity

-  A single weak learner is usually simple, rigid, and exhibits high bias (underfitting) because it cannot capture complex patterns in the dat
-  Instead of training models independently, boosting works additively and sequentially. Model 2 is trained to predict the residual errors or mistakes left behind by Model 1.


## AdaBoost

The central idea is simple:
Train a sequence of weak models, give more attention to examples that previous models got wrong, and combine all models using a weighted vote.

### how inferencing happens in gradient boosting

In Gradient Boosting algorithms (GBDT, XGBoost, LightGBM, CatBoost), inference happens by passing a new observation through all the trained trees, collecting their leaf values, and adding those values to the initial prediction.
The important point: During inference, we do not calculate gradients, Hessians, residuals, or update trees. Those operations happen during training.
