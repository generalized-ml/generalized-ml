```python
# lets simulate central limit thoerem with python


import matplotlib.pyplot as plt

#define a random variable with a uniform distribution
import numpy as np
samples = 10000
#keep the plots different for both the variable


X = np.random.uniform(0, 1, samples)
means = []
for i in range(1000):
    means.append(np.mean(np.random.choice(X, 100)))


plt.hist(X, bins=30, density=True)
plt.title('Uniform Distribution')
plt.xlabel('Value')
plt.ylabel('Density')
plt.show()
# now take the mean of 100 samples of X and plot the histogram

plt.hist(means, bins=30, density=True)
plt.title('Distribution of Sample Means')
plt.xlabel('Sample Mean')
plt.ylabel('Density')
plt.show()
```

So this means any mean of iid will have normal ditribution if we take mean several times

- Law of Large Numbers: What happens to the sample mean as the sample size grows?
The most important takeaway is that the CLT lets us use the mathematics of the normal distribution to perform statistical inference even when the underlying population is not normally distributed, subject to its assumptions.

## Central Limit Theorem (CLT)

For **independent, identically distributed** observations with **finite variance**, the sampling distribution of the sample mean approaches a normal distribution as $n$ increases:

$$
\boxed{\bar{X} \approx N\left(\mu,\ \frac{\sigma^2}{n}\right)}
$$

### Notation

| Symbol | Meaning |
|:------:|:--------|
| $\bar{X}$ | Sample mean |
| $\mu$ | Population mean |
| $\sigma$ | Population standard deviation |
| $n$ | Sample size |
| $\dfrac{\sigma}{\sqrt{n}}$ | Standard error of the sample mean |

### Key Results

$$
E[\bar{X}] = \mu
$$

$$
SE(\bar{X}) = \frac{\sigma}{\sqrt{n}}
$$

> **Takeaway:** The sample means are centered around the population mean, and their variability **decreases** as the sample size increases.
